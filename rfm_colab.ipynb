```python
import os, sys
if not os.path.exists("data-analyst-portfolio"):
    !git clone -q https://github.com/oum292929/data-analyst-portfolio.git
%cd data-analyst-portfolio/01-marketing-rfm-segmentation
sys.path.insert(0, "src")

import pandas as pd
import rfm_analysis as rfm
from IPython.display import display, Image
```

## 2. Load the data

**Option A** downloads the dataset from Kaggle automatically. If it asks you to log in or fails, use **Option B**.

```python
# Option A: download from Kaggle
!pip -q install kagglehub
import kagglehub, glob
folder = kagglehub.dataset_download("limjeongeun/synthetic-u-s-e-commerce-dataset-1m-orders")
csv_files = glob.glob(os.path.join(folder, "**", "*.csv"), recursive=True)
print("CSV files found:")
for f in csv_files:
    print(" ", f, f"({os.path.getsize(f)/1e6:.1f} MB)")
CSV_PATH = csv_files[0]   # if there are several files, set this to the orders file
print("Using:", CSV_PATH)
```

```python
# Option B: upload the CSV yourself (run this cell only if Option A did not work)
from google.colab import files
uploaded = files.upload()
CSV_PATH = list(uploaded.keys())[0]
print("Using:", CSV_PATH)
```

## 3. Inspect the raw data

```python
raw = pd.read_csv(CSV_PATH, low_memory=False)
print(raw.shape)
display(raw.head())
display(raw.dtypes)
```

## 4. Data cleaning

The cleaning log records the number of rows after each step. Use it to fill the Data Cleaning table in the README.

```python
pd.set_option("display.max_colwidth", 120)
data, cleaning_log = rfm.clean(raw)
display(cleaning_log)
os.makedirs("images", exist_ok=True)
cleaning_log.to_csv("images/cleaning_log.csv", index=False)
```

## 5. RFM segmentation

```python
rfm_table = rfm.build_rfm(data)
summary = rfm.summarize(rfm_table)
display(summary)
summary.to_csv("images/segment_summary.csv")
rfm.plot(summary, rfm.OUT_DIR / "rfm_segments.png")
display(Image("images/rfm_segments.png"))
```
